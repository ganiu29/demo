```python

BASE = r"c:\Users\ainag\OneDrive - Datore\Rapport\Position Category"
THRESHOLD = 0.90

ref = pd.read_csv(f"{BASE}\\position_categorised.csv")
target = pd.read_csv(f"{BASE}\\position.csv")


def normalise(title):
    # Lower-case and collapse whitespace so case or spacing differences don't lower the score
    return " ".join(str(title).lower().split())


ref["_key"] = ref["Job Title"].map(normalise)
ref_keys = ref["_key"].tolist()
ref_categories = ref["Category"].tolist()


def match_category(title):
    key = normalise(title)
    best_score, best_category = 0.0, None
    for ref_key, category in zip(ref_keys, ref_categories):
        score = difflib.SequenceMatcher(None, key, ref_key).ratio()
        if score > best_score:
            best_score, best_category = score, category
    return best_category if best_score >= THRESHOLD else None


target["Category"] = target["Job Title"].map(match_category)
target.to_csv(f"{BASE}\\position_matched.csv", index=False)

print(f"Matched {target['Category'].notna().sum()} of {len(target)} job titles")
```
